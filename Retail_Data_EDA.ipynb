```python
retail.nlargest(5, "Profit")[["Customer", "Sales", "Profit", "Profit_Margin"]]
# this shows a issue in the data.
# for customer C024, we see an issue in the Profit calculation.
```

```python
retail["Expected_Profit"] = (
    retail["Sales"] * retail["Profit_Margin"] / 100
)

retail["Profit_Difference"] = (
    retail["Profit"] - retail["Expected_Profit"]
)
```

```python
retail.nlargest(5, "Profit_Difference")[
    ["Customer", "Sales", "Profit", "Profit_Margin",
     "Expected_Profit", "Profit_Difference"]
]
# this shows a difference in the profit values after the calculation.
```

```python
retail["Calculated_Margin"] = (
    retail["Profit"] / retail["Sales"] * 100
)

retail["Margin_Difference"] = (
    retail["Calculated_Margin"] - retail["Profit_Margin"]
)
```

```python
retail.nlargest(5, "Margin_Difference")[
    ["Customer", "Sales", "Profit", "Profit_Margin",
     "Calculated_Margin", "Margin_Difference"]
]
```

```python
retail.nsmallest(5, "Margin_Difference")[
    ["Customer", "Sales", "Profit", "Profit_Margin",
     "Calculated_Margin", "Margin_Difference"]
]
```

```python
# we found the issue in the profit calculation in the top 5 and bottom 5 values so we have to check the entire for the issue.
retail[retail["Profit_Difference"].abs() > 10][
    ["Customer", "Sales", "Profit", "Profit_Margin",
     "Expected_Profit", "Profit_Difference"]
]
```

```python
#since we found the issue in only these 3 rows lets calculate the profit for these 3 rows and replace it.
retail.loc[retail["Customer"].isin(["C024", "C015", "C059"]), "Profit"] = (
    retail.loc[retail["Customer"].isin(["C024", "C015", "C059"]), "Sales"]
    * retail.loc[retail["Customer"].isin(["C024", "C015", "C059"]), "Profit_Margin"]
    / 100
)
```

```python
retail.nlargest(5, "Margin_Difference")[
    ["Customer", "Sales", "Profit", "Profit_Margin",
     "Calculated_Margin", "Margin_Difference"]
]
```

```python
retail.nsmallest(5, "Margin_Difference")[
    ["Customer", "Sales", "Profit", "Profit_Margin",
     "Calculated_Margin", "Margin_Difference"]
]
```

```python
# as the profit calculation corrected now, lets drop the diagnostic columns
retail = retail.drop(columns=[
    "Expected_Profit",
    "Profit_Difference",
    "Calculated_Margin",
    "Margin_Difference"
])
```
